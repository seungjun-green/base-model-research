```python
# Set up experiment runtime
"""Standard-library setup embedded in the notebooks so it can run before the checkout exists."""

import fcntl
import json
import re
import subprocess
import sys
from importlib.metadata import PackageNotFoundError, version

GITHUB_REPO_URL = "https://github.com/seungjun-green/base-model-research"


def _checkout(workspace, runs):
    source_file = runs / "source.json"
    if source_file.exists():
        source = json.loads(source_file.read_text())
        if source["url"].rstrip("/").removesuffix(".git").lower() != GITHUB_REPO_URL.lower():
            raise ValueError(
                f"{runs} is already pinned to a different repository. Use a new EXPERIMENT name."
            )
        commit = source["commit"]
    else:
        head = subprocess.check_output(
            ["git", "ls-remote", "--exit-code", GITHUB_REPO_URL, "HEAD"], text=True
        )
        commit = head.split()[0]
        source = {"url": GITHUB_REPO_URL, "commit": commit}
    if not re.fullmatch(r"[0-9a-f]{40}", commit):
        raise ValueError(f"Invalid Git commit for this experiment: {commit!r}")

    repo = workspace / "repos" / commit
    if not repo.exists():
        repo.parent.mkdir(parents=True, exist_ok=True)
        subprocess.run(["git", "clone", "--", GITHUB_REPO_URL, repo], check=True)
    if subprocess.check_output(["git", "status", "--porcelain"], cwd=repo, text=True).strip():
        raise RuntimeError(f"The runtime checkout has local edits: {repo}")
    if subprocess.check_output(["git", "rev-parse", "HEAD"], cwd=repo, text=True).strip() != commit:
        subprocess.run(["git", "fetch", "origin", commit], cwd=repo, check=True)
        subprocess.run(["git", "checkout", "--detach", commit], cwd=repo, check=True)
    if not source_file.exists():
        temporary = source_file.with_suffix(".tmp")
        temporary.write_text(json.dumps(source, indent=2) + "\n")
        temporary.replace(source_file)
    return repo


def setup_colab(workspace, runs, *, benchmark):
    runs.mkdir(parents=True, exist_ok=True)
    with (runs / ".setup.lock").open("a") as lock:
        fcntl.flock(lock, fcntl.LOCK_EX)
        repo = _checkout(workspace, runs)
        try:
            torch_version = version("torch")
        except PackageNotFoundError as error:
            raise RuntimeError("Select a Colab runtime with PyTorch installed and rerun setup.") from error

        # Keep the runtime's PyTorch build, including its CUDA version suffix.
        extras = "data,eval" if benchmark else "data"
        print(f"Using {sys.executable} with PyTorch {torch_version}", flush=True)
        subprocess.run(
            [
                sys.executable,
                "-m",
                "pip",
                "install",
                "--quiet",
                f"torch=={torch_version}",
                "-e",
                f"{repo}[{extras}]",
            ],
            check=True,
        )
    return repo

REPO = setup_colab(WORKSPACE, RUNS, benchmark=False)

sys.path.insert(0, str(REPO / "src"))
from base_model_research.progress import run_with_progress

"""Proxy notebook helpers, embedded so they work with an existing experiment's pinned trainer."""

import json
import shutil
from pathlib import Path

import torch
from tqdm.auto import tqdm

from base_model_research.config import ModelConfig, TrainConfig, dump_json
from base_model_research.data import TokenCorpus, file_sha256
from base_model_research.diagnostics import DEFAULT_HP
from base_model_research.models import ResearchLM, parameter_count
from base_model_research.storage import run_lock


def stage_proxy_data(source, data, training_targets):
    """Stage the training prefix (including lookahead) and every validation shard."""
    manifest = source / "manifest.json"
    if not manifest.is_file():
        raise FileNotFoundError(f"Upload the prepared FineWeb-Edu folder to {source} before training.")
    local_manifest = data / "manifest.json"
    if source == data or (local_manifest.is_file() and local_manifest.read_bytes() == manifest.read_bytes()):
        print(f"Using the prepared local dataset at {data}.", flush=True)
        return data

    train = TokenCorpus(source, "train")
    validation = TokenCorpus(source, "validation")
    if train.total_tokens < training_targets + 1:
        raise ValueError(f"This proxy needs {training_targets + 1:,} training tokens including lookahead.")
    local = data.parent / "proxy_shards" / train.fingerprint
    with run_lock(local, blocking=True):
        required = []
        for corpus, limit in ((train, training_targets + 1), (validation, validation.total_tokens)):
            for start, entry in zip(corpus.starts, corpus.entries):
                name = entry["path"]
                if Path(name).name != name or name in ("", ".", ".."):
                    raise ValueError(f"Expected a shard filename: {name!r}")
                if start < limit:
                    required.append(entry)
                elif not (local / name).exists():
                    # Pinned trainers stat every shard listed in the original manifest.
                    # Only shards beyond this proxy's training budget are links to Drive.
                    (local / name).symlink_to(source / name)

        print(f"Preparing {sum(e['tokens'] * 4 for e in required) / 1e9:.2f} GB at {local}", flush=True)
        for entry in tqdm(required, desc="Copying or verifying proxy shards", unit="shard"):
            target = local / entry["path"]
            if not target.is_symlink() and target.is_file() and file_sha256(target) == entry["sha256"]:
                continue
            temporary = target.with_suffix(".copying")
            try:
                shutil.copyfile(source / entry["path"], temporary)
                if file_sha256(temporary) != entry["sha256"]:
                    raise ValueError(f"Shard checksum mismatch: {entry['path']}")
                temporary.replace(target)
            finally:
                temporary.unlink(missing_ok=True)
        # Keep the full corpus identity for saved trials and the main-model transfer.
        dump_json(local / "manifest.json", train.manifest)
    return local


def proxy_training_targets(repo, family):
    model = ModelConfig.load(repo / "configs" / "models" / f"{family}_proxy.yaml")
    config = TrainConfig.load(repo / "configs" / "training.yaml")
    with torch.device("meta"):
        parameters = parameter_count(ResearchLM(model, DEFAULT_HP))
    return parameters * config.tokens_per_parameter


def proxy_sweep_args(repo, runs, family, microbatch_size, local, *, reuse_trial_000=None):
    model_name = f"{family}_proxy"
    model_path = repo / "configs" / "models" / f"{model_name}.yaml"
    protocol = runs / "proxy" / ModelConfig.load(model_path).family / "protocol.json"
    settings_path = runs / f"{model_name}_profile.json"
    if reuse_trial_000 is not None and (reuse_trial_000 / "run.json").is_file():
        settings_path = reuse_trial_000 / "run.json"
    if protocol.is_file():
        settings_path = protocol
    settings = json.loads(settings_path.read_text())["training"] if settings_path.is_file() else {}
    if microbatch_size is None and reuse_trial_000 is not None:
        microbatch_size = settings.get("microbatch_size", 16)
    if type(microbatch_size) is not int:
        raise ValueError("MICROBATCH_SIZE must be a positive integer that divides the global batch size.")
    config = TrainConfig.load(repo / "configs" / "training.yaml", microbatch_size=microbatch_size)
    if settings_path == protocol and settings["microbatch_size"] != microbatch_size:
        raise ValueError(
            f"This {family} sweep uses MICROBATCH_SIZE = {settings['microbatch_size']}. "
            f"Use that value to continue, or set a new EXPERIMENT name to start a sweep at {microbatch_size}."
        )
    accumulation = config.global_batch_size // microbatch_size
    print(f"{model_name}: {accumulation} microbatches of {microbatch_size} sequences per update.", flush=True)
    args = [
        "--model",
        str(model_path),
        "--data",
        str(local),
        "--output",
        str(runs / "proxy"),
        "--microbatch-size",
        str(microbatch_size),
        "--progress-events",
    ]
    if settings.get("activation_checkpointing", config.activation_checkpointing):
        args.append("--activation-checkpointing")
    if reuse_trial_000 is not None:
        args.extend(["--reuse-trial-000", str(reuse_trial_000)])
    return args
```

Each proxy uses **trial 000**, the settings under which the original gated proxy diverged.
All three use seed 42, the same ordered TinyLlama-tokenized FineWeb-Edu data, and the same
held-out validation set. They run sequentially for 20 training tokens per parameter with
20 validations per full run. The parameter counts and total updates can differ.

In the table, `x` is the input to the sublayer, `y` its output, and `p` the previous
layer's output from the **same kind of sublayer**. Multiplication `⊙` is elementwise.

| Variant | Output passed to the next sublayer |
|---|---|
| 1 · add_previous | GELU(W₁x) ⊙ (W₂y) + p |
| 2 · scale_previous | GELU(W₁x) ⊙ (W₂y) + αp |
| 3 · multiply_previous | GELU(ReLU(x + y)) ⊙ (Wp) |

The first layer uses ordinary residual addition because it has no previous layer.
Every later layer applies its formula to both attention and the FFN, with independent
weights. Each α in variant 2 is a learned scalar initialized to 1.


```python
from dataclasses import replace
from base_model_research.config import DEFAULT_TOKENIZER, GATE_VARIANTS, Hyperparameters
from base_model_research.search import load_trials

trial = next(t for t in load_trials(REPO / "experiments/trials_seed42.json")["trials"]
             if t["id"] == "trial_000")
hp = Hyperparameters(**trial["hyperparameters"])
settings = TrainConfig.load(REPO / "configs/training.yaml", microbatch_size=MICROBATCH_SIZE)
proxy = ModelConfig.load(REPO / "configs/models/modified_main.yaml").at_width(1)
models = [replace(proxy, family=name) for name in GATE_VARIANTS]
with torch.device("meta"):
    counts = {m.family: parameter_count(ResearchLM(m, hp)) for m in models}
print(f"Shared settings: {trial['id']}, seed {settings.seed}, microbatch {settings.microbatch_size}")
print(json.dumps(trial["hyperparameters"], indent=2))
for name, count in counts.items():
    print(f"{name}: {count:,} parameters, {count * settings.tokens_per_parameter:,} training tokens")
```

```python
# Copy the shards needed by the largest variant to the Colab SSD.
manifest = SAVED_DATA / "manifest.json"
if not manifest.is_file():
    raise FileNotFoundError(f"Expected your prepared TinyLlama corpus at {manifest}.")
if json.loads(manifest.read_text())["tokenizer"] != DEFAULT_TOKENIZER:
    raise ValueError(f"Expected the TinyLlama-tokenized FineWeb-Edu corpus at {SAVED_DATA}.")
PROXY_DATA = stage_proxy_data(SAVED_DATA, DATA, max(counts.values()) * settings.tokens_per_parameter)
```

```python
from contextlib import chdir
from math import ceil
from base_model_research.training import train

def train_variant(model):
    output = RUNS / model.family / trial["id"]
    total = ceil(counts[model.family] * settings.tokens_per_parameter /
                 (settings.global_batch_size * settings.context_length))
    print(f"Checking saved results and initializing {model.family}…", flush=True)
    with chdir(REPO), tqdm(total=total, desc=model.family, unit="update", delay=1) as bar:
        def show_update(label, unit, completed, target, detail=""):
            bar.set_postfix_str(detail, refresh=False)
            bar.update(completed - bar.n)

        return train(model, hp, settings, PROXY_DATA, output,
                     resume=(output / "run.json").is_file(),
                     selection={"trial_id": trial["id"]}, progress=show_update)

for model in models:
    result = train_variant(model)
    print(f"{model.family}: {result['status']} after {result['step']:,} completed updates.")
    if result["status"] == "diverged":
        print(f"Update {result['failed_at_step']}: {result['failure_reason']}.")
    if result["status"] == "interrupted":
        break
```

**Read the results.** A nonfinite training loss, gradient norm, or validation loss marks
that variant as diverged. The next variant then starts. Compare final validation losses
among completed runs. A loss near 10.37 is what uniform predictions give with this
32,000-token vocabulary, so a finite loss alone can hide a model that learned very little.
Watch the gradient norm in the progress bar. If it stays at zero while loss stays near
10.37, the model may be stuck predicting every token with equal probability.

Rerunning the training cell reuses completed and diverged results. An interrupted variant
restarts from its seed. You can run the summary cell separately to inspect saved results.


```python
import pandas as pd

rows = []
for model in models:
    folder = RUNS / model.family / trial["id"]
    path = folder / "result.json"
    result = json.loads(path.read_text()) if path.is_file() else {}
    metrics = folder / "metrics.jsonl"
    logged = metrics.read_text().splitlines() if result and metrics.is_file() else []
    last = json.loads(logged[-1]) if logged else {}
    rows.append({
        "variant": model.family, "parameters": counts[model.family],
        "status": result.get("status", "no result yet"),
        "completed_updates": result.get("step"), "failed_at_update": result.get("failed_at_step"),
        "failure_reason": result.get("failure_reason"), "training_tokens": result.get("consumed_tokens"),
        "last_logged_gradient_norm": last.get("gradient_norm"),
        "final_validation_loss": result.get("validation_loss"),
    })
summary = pd.DataFrame(rows).convert_dtypes()
summary_path = RUNS / "comparison.csv"
temporary = summary_path.with_suffix(".csv.tmp")
with run_lock(RUNS, blocking=True):
    summary.to_csv(temporary, index=False)
    temporary.replace(summary_path)
print(f"Saved comparison: {summary_path}")
summary
```
